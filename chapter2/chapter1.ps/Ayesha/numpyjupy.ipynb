```python
import time
my_list = list(range(100000000))
start = time.time()
my_list = [x+1 for x in my_list]
print(f"python list took:{time.time()-start:4f}seconds")
```

```python
import numpy as np
my_array = np.arange(100000000)
start = time.time()
my_array = my_array+1
print(f"NumPy took: {time.time()-start:4f}seconds")
```
